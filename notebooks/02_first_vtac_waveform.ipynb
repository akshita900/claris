```python
import wfdb
import matplotlib.pyplot as plt

record_path = "../data/raw/vtac/003c13_0115"

record = wfdb.rdrecord(record_path)

print("Signals:", record.n_sig)
print("Sampling frequency:", record.fs)
print("Samples:", record.sig_len)
print("Signal names:", record.sig_name)
```

```python
fig, axes = plt.subplots(record.n_sig, 1, figsize=(16, 10), sharex=True)

time = record.p_signal.shape[0] / record.fs

for i, ax in enumerate(axes):
    ax.plot(
        record.p_signal[:, i],
        linewidth=0.7
    )
    ax.set_ylabel(record.sig_name[i])
    ax.grid(True)

axes[-1].set_xlabel("Sample")
fig.suptitle("VTaC Event 003c13_0115 — Full 6-Minute Waveform")

plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

fs = record.fs

# Time axis in seconds
time = np.arange(record.sig_len) / fs

# Alarm onset
alarm_time = 5 * 60  # 5 minutes
alarm_sample = int(alarm_time * fs)

fig, axes = plt.subplots(
    record.n_sig,
    1,
    figsize=(16, 10),
    sharex=True
)

for i, ax in enumerate(axes):
    ax.plot(time, record.p_signal[:, i], linewidth=0.7)
    
    # Mark VT alarm onset
    ax.axvline(
        alarm_time,
        linestyle="--",
        linewidth=1.5
    )
    
    ax.set_ylabel(record.sig_name[i])
    ax.grid(True)

axes[-1].set_xlabel("Time (seconds)")

fig.suptitle(
    "VTaC Event 003c13_0115 — Alarm Onset at 5 Minutes"
)

plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Window around the alarm
start_time = 240
end_time = 330

start_sample = int(start_time * record.fs)
end_sample = int(end_time * record.fs)

time_window = np.arange(start_sample, end_sample) / record.fs

fig, axes = plt.subplots(
    record.n_sig,
    1,
    figsize=(16, 10),
    sharex=True
)

for i, ax in enumerate(axes):
    ax.plot(
        time_window,
        record.p_signal[start_sample:end_sample, i],
        linewidth=0.7
    )

    ax.axvline(
        300,
        linestyle="--",
        linewidth=1.5
    )

    ax.set_ylabel(record.sig_name[i])
    ax.grid(True)

axes[-1].set_xlabel("Time (seconds)")

fig.suptitle(
    "VTaC Event 003c13_0115 — 60 s Before / 30 s After Alarm"
)

plt.tight_layout()
plt.show()
```
